### Absorptionsspektra:

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df = pd.read_csv("/home/daniel/Desktop/Uni/6 Semester/Bachelorprojekt/bachelorprojekt/Data/Absorptiondata.txt", sep='\t')

lam = df["250"]
HbO2 = df["106112"]
Hb = df["112736"]

plt.figure(figsize=(13,5), dpi=300)
plt.semilogy(lam, HbO2, color="red", label="HbO$_2$")
plt.semilogy(lam, Hb, color="blue", label="Hb")
plt.grid()
plt.xlabel("Bølgelængde [nm]")
plt.ylabel(r"Molar ekstinktionskoefficient $\varepsilon$ [cm$^{-1}$ M$^{-1}$]")
plt.axvline(x=800, color="green", linestyle="dashed")
plt.xticks([300, 400, 500, 600, 700, 800, 900, 1000])

plt.legend()
plt.savefig("/home/daniel/Desktop/Uni/6 Semester/Bachelorprojekt/bachelorprojekt/Data/Plots/extinction_spectra.pdf")
plt.show()
```
