### Appendix: Sammenligning af LED-spektre


```python
import matplotlib.pyplot as plt
import pandas as pd

base_path = "/home/daniel/Desktop/Uni/6 Semester/Bachelorprojekt/bachelorprojekt/data/raw/daniel/leder"
hot_pixel_nm = 996.65
hot_pixel_width_nm = 2

spectra = {
    "660 nm": ("VIS_660NM.txt", 660, "red"),
    "810 nm": ("IR_810NM.txt", 810, "tab:orange"),
    "940 nm": ("IR_940NM.txt", 940, "tab:blue"),
    "970 nm": ("IR_970NM.txt", 970, "tab:purple"),
}

plt.figure(figsize=(13, 5), dpi=300)

for label, (filename, nominal_wavelength, color) in spectra.items():
    df = pd.read_csv(
        f"{base_path}/{filename}",
        sep="\t",
        skiprows=17,
        skipfooter=1,
        engine="python",
        decimal=",",
        names=["Bølgelængde", "Intensitet"],
    )

    df = df[~df["Bølgelængde"].between(hot_pixel_nm - hot_pixel_width_nm, hot_pixel_nm + hot_pixel_width_nm)]
    window = df[df["Bølgelængde"].between(nominal_wavelength - 80, nominal_wavelength + 80)].copy()
    baseline = window["Intensitet"].min()
    peak = window.loc[window["Intensitet"].idxmax()]
    half_max = baseline + (peak["Intensitet"] - baseline) / 2

    above_half = window[window["Intensitet"] >= half_max]
    left_i = above_half.index[0]
    right_i = above_half.index[-1]

    left_pos = window.index.get_loc(left_i)
    right_pos = window.index.get_loc(right_i)

    if left_pos > 0:
        left_before_i = window.index[left_pos - 1]
        x1, y1 = window.loc[left_before_i, ["Bølgelængde", "Intensitet"]]
        x2, y2 = window.loc[left_i, ["Bølgelængde", "Intensitet"]]
        left_half = x1 + (half_max - y1) * (x2 - x1) / (y2 - y1)
    else:
        left_half = window.loc[left_i, "Bølgelængde"]

    if right_pos < len(window) - 1:
        right_after_i = window.index[right_pos + 1]
        x1, y1 = window.loc[right_i, ["Bølgelængde", "Intensitet"]]
        x2, y2 = window.loc[right_after_i, ["Bølgelængde", "Intensitet"]]
        right_half = x1 + (half_max - y1) * (x2 - x1) / (y2 - y1)
    else:
        right_half = window.loc[right_i, "Bølgelængde"]

    fwhm = right_half - left_half
    normalized_intensity = df["Intensitet"] / peak["Intensitet"]
    normalized_half_max = half_max / peak["Intensitet"]

    plt.plot(df["Bølgelængde"], normalized_intensity, color=color, label=f"{label}, FWHM = {fwhm:.1f} nm")
    plt.hlines(normalized_half_max, left_half, right_half, color=color, linestyle="dotted", linewidth=2)
    plt.axvline(nominal_wavelength, color=color, linestyle="dashed", alpha=0.45)

plt.grid(alpha=0.3)
plt.xlabel("Bølgelængde [nm]")
plt.ylabel("Normaliseret intensitet")
plt.xlim((600, 1020))
plt.ylim((-0.05, 1.25))
plt.legend(title="Nominel LED")
plt.tight_layout()
plt.savefig("/home/daniel/Desktop/Uni/6 Semester/Bachelorprojekt/bachelorprojekt/data/plots/led_spectra_normalized_comparison.pdf", bbox_inches="tight")
plt.show()
```

```python
import pandas as pd

base_path = "/home/daniel/Desktop/Uni/6 Semester/Bachelorprojekt/bachelorprojekt/data/raw/daniel/leder"
hot_pixel_nm = 996.65
hot_pixel_width_nm = 2

spectra = {
    "660 nm": ("VIS_660NM.txt", 660),
    "810 nm": ("IR_810NM.txt", 810),
    "940 nm": ("IR_940NM.txt", 940),
    "970 nm": ("IR_970NM.txt", 970),
}

rows = []
for label, (filename, nominal_wavelength) in spectra.items():
    df = pd.read_csv(
        f"{base_path}/{filename}",
        sep="\t",
        skiprows=17,
        skipfooter=1,
        engine="python",
        decimal=",",
        names=["Bølgelængde", "Intensitet"],
    )

    df = df[~df["Bølgelængde"].between(hot_pixel_nm - hot_pixel_width_nm, hot_pixel_nm + hot_pixel_width_nm)]
    window = df[df["Bølgelængde"].between(nominal_wavelength - 80, nominal_wavelength + 80)].copy()
    baseline = window["Intensitet"].min()
    peak = window.loc[window["Intensitet"].idxmax()]
    half_max = baseline + (peak["Intensitet"] - baseline) / 2

    above_half = window[window["Intensitet"] >= half_max]
    left_i = above_half.index[0]
    right_i = above_half.index[-1]

    left_pos = window.index.get_loc(left_i)
    right_pos = window.index.get_loc(right_i)

    if left_pos > 0:
        left_before_i = window.index[left_pos - 1]
        x1, y1 = window.loc[left_before_i, ["Bølgelængde", "Intensitet"]]
        x2, y2 = window.loc[left_i, ["Bølgelængde", "Intensitet"]]
        left_half = x1 + (half_max - y1) * (x2 - x1) / (y2 - y1)
    else:
        left_half = window.loc[left_i, "Bølgelængde"]

    if right_pos < len(window) - 1:
        right_after_i = window.index[right_pos + 1]
        x1, y1 = window.loc[right_i, ["Bølgelængde", "Intensitet"]]
        x2, y2 = window.loc[right_after_i, ["Bølgelængde", "Intensitet"]]
        right_half = x1 + (half_max - y1) * (x2 - x1) / (y2 - y1)
    else:
        right_half = window.loc[right_i, "Bølgelængde"]

    rows.append({
        "LED": label,
        "Nominel bølgelængde [nm]": nominal_wavelength,
        "Målt peak [nm]": round(peak["Bølgelængde"], 2),
        "Afvigelse [nm]": round(peak["Bølgelængde"] - nominal_wavelength, 2),
        "FWHM [nm]": round(right_half - left_half, 2),
        "Maks intensitet [counts]": round(peak["Intensitet"], 2),
    })

peak_table = pd.DataFrame(rows)
peak_table
```
