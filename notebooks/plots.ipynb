### Pulsoximeter LED

```python
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path

repo_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
path = repo_root / "data/raw/daniel/lysdioder/PULSOXIMETER-BRAUN.txt"
plot_dir = repo_root / "data/plots"
plot_dir.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(
    path,
    sep="\t",
    skiprows=17,
    skipfooter=1,
    engine="python",
    decimal=",",
    names=["Bølgelængde", "Intensitet"],
)

window = df[df["Bølgelængde"].between(620, 700)].copy()
baseline = window["Intensitet"].min()
peak = window.loc[window["Intensitet"].idxmax()]
half_max = baseline + (peak["Intensitet"] - baseline) / 2

above_half = window[window["Intensitet"] >= half_max]
left_i = above_half.index[0]
right_i = above_half.index[-1]
left_pos = window.index.get_loc(left_i)
right_pos = window.index.get_loc(right_i)

if left_pos > 0:
    left_before_i = window.index[left_pos - 1]
    x1, y1 = window.loc[left_before_i, ["Bølgelængde", "Intensitet"]]
    x2, y2 = window.loc[left_i, ["Bølgelængde", "Intensitet"]]
    left_half = x1 + (half_max - y1) * (x2 - x1) / (y2 - y1)
else:
    left_half = window.loc[left_i, "Bølgelængde"]

if right_pos < len(window) - 1:
    right_after_i = window.index[right_pos + 1]
    x1, y1 = window.loc[right_i, ["Bølgelængde", "Intensitet"]]
    x2, y2 = window.loc[right_after_i, ["Bølgelængde", "Intensitet"]]
    right_half = x1 + (half_max - y1) * (x2 - x1) / (y2 - y1)
else:
    right_half = window.loc[right_i, "Bølgelængde"]

fwhm = right_half - left_half
normalized_intensity = (window["Intensitet"] - baseline) / (peak["Intensitet"] - baseline)

fig, ax = plt.subplots(figsize=(13, 5), dpi=300, constrained_layout=True)
ax.plot(
    window["Bølgelængde"],
    normalized_intensity,
    color="tab:red",
    linewidth=1.6,
    label=f"Kommercielt Braun pulsoximeter, FWHM = {fwhm:.1f} nm",
)
ax.hlines(0.5, left_half, right_half, color="black", linestyle="dotted", linewidth=1.8, label="Halv maksimum")
ax.axvline(peak["Bølgelængde"], color="tab:red", linestyle="dashed", alpha=0.65, linewidth=1.4)
ax.scatter([peak["Bølgelængde"]], [1.0], color="tab:red", edgecolor="black", linewidth=0.6, zorder=3)
ax.text(
    0.03,
    0.94,
    f"Peak: {peak['Bølgelængde']:.1f} nm\nFWHM: {fwhm:.1f} nm",
    transform=ax.transAxes,
    va="top",
    ha="left",
    bbox={"boxstyle": "round,pad=0.35", "facecolor": "white", "edgecolor": "0.75", "alpha": 0.95},
)
ax.grid(alpha=0.3)
ax.set_xlabel("Bølgelængde [nm]")
ax.set_ylabel("Normaliseret intensitet")
ax.set_xlim((620, 700))
ax.set_ylim((-0.05, 1.15))
ax.legend(loc="upper right")
fig.savefig(plot_dir / "Braun_pulsoximeter.pdf", bbox_inches="tight")
fig.savefig(plot_dir / "Braun_pulsoximeter.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.ticker import LogLocator, NullFormatter
from pathlib import Path

repo_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
path = repo_root / "data/raw/daniel/waveforms scope/LPF BODE.csv"
plot_dir = repo_root / "data/plots"
plot_dir.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(path, comment="#")
frequency = df["Frequency (Hz)"].to_numpy(dtype=float)
magnitude_db = df["Channel 1 Magnitude (dB)"].to_numpy(dtype=float)

passband_db = np.median(magnitude_db[:max(5, len(magnitude_db) // 10)])
cutoff_db = passband_db - 3.0
crossings = np.flatnonzero((magnitude_db[:-1] >= cutoff_db) & (magnitude_db[1:] <= cutoff_db))
if len(crossings) == 0:
    cutoff_index = int(np.argmin(np.abs(magnitude_db - cutoff_db)))
    cutoff_frequency = frequency[cutoff_index]
else:
    cutoff_index = crossings[0]
    f0, f1 = frequency[cutoff_index], frequency[cutoff_index + 1]
    y0, y1 = magnitude_db[cutoff_index], magnitude_db[cutoff_index + 1]
    log_f0, log_f1 = np.log10(f0), np.log10(f1)
    cutoff_frequency = 10 ** (log_f0 + (cutoff_db - y0) * (log_f1 - log_f0) / (y1 - y0))

plt.rcParams.update({
    "font.size": 10,
    "axes.labelsize": 10,
    "axes.titlesize": 11,
    "legend.fontsize": 9,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})

fig, ax = plt.subplots(figsize=(13, 5), dpi=300)
ax.semilogx(
    frequency,
    magnitude_db,
    color="#b15f00",
    linewidth=2.0,
    solid_capstyle="round",
    label="Målt frekvensrespons",
)
ax.axhline(cutoff_db, color="0.28", linestyle=(0, (4, 3)), linewidth=1.0, label="-3 dB niveau")
ax.axvline(cutoff_frequency, color="0.12", linestyle=(0, (2, 2)), linewidth=1.0)
ax.scatter(
    [cutoff_frequency],
    [cutoff_db],
    s=32,
    color="white",
    edgecolor="0.12",
    linewidth=1.1,
    zorder=4,
)
ax.annotate(
    f"$f_c$ = {cutoff_frequency:.2g} Hz",
    xy=(cutoff_frequency, cutoff_db),
    xycoords="data",
    xytext=(0.58, 0.18),
    textcoords="axes fraction",
    arrowprops=dict(arrowstyle="->", color="0.18", linewidth=0.9, shrinkA=3, shrinkB=4),
    bbox=dict(boxstyle="round,pad=0.24", facecolor="white", edgecolor="0.75", linewidth=0.8),
)
ax.set_title("Lavpasfilter: målt frekvensrespons")
ax.set_xlabel("Frekvens [Hz]")
ax.set_ylabel("Amplitude [dB]")
ax.set_xlim(frequency.min(), frequency.max())
y_margin = 1.5
ax.set_ylim(magnitude_db.min() - y_margin, magnitude_db.max() + y_margin)
ax.xaxis.set_major_locator(LogLocator(base=10))
ax.xaxis.set_minor_locator(LogLocator(base=10, subs=np.arange(2, 10) * 0.1))
ax.xaxis.set_minor_formatter(NullFormatter())
ax.grid(True, which="major", color="0.78", linewidth=0.75)
ax.grid(True, which="minor", axis="x", color="0.88", linewidth=0.45)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.legend(loc="best", frameon=False)
fig.tight_layout(pad=0.8)
fig.savefig(plot_dir / "lpf_bode_scope.png", bbox_inches="tight", dpi=300)
plt.show()
```
